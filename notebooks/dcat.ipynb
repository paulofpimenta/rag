`````python
from langchain_community.vectorstores import Neo4jVector
import os
from dotenv import find_dotenv
from langchain_ollama import OllamaEmbeddings
import json
from langchain_core.documents import Document
from langchain_community.vectorstores import Neo4jVector

find_dotenv()

NEO4J_URL = "bolt://localhost:7687"
NEO4J_USER = "neo4j"
NEO4J_PASSWORD = os.environ.get('NEO4JPASS')
hugging_face_token = os.environ.get('HUGGING_FACE')


# 1. Load the PubMed Validation subset (3% of 133,000)


REPO_ID = "scientific_papers"
FILEPATH = "../data/pubmed_val.txt"
INDEX_NAME = "pubmed_description_index"


# 2. Initialize the embedding model specified in the paper
ollama_embeddings = OllamaEmbeddings(model="nomic-embed-text")

# This matches the study's source for dynamic few-shot examples
# Old dataset, must donwload direclly
print(" Manually loading and ingesting PubMed validation subset...")


# 4. Prepare documents for ingestion
def ingest_pubmed_in_batches(file_path, batch_size=100):
    batch = []
    vector_db = None
    total_count = 0

    print(f"Starting ingestion from {file_path}...")

    with open(file_path, 'r', encoding='utf-8') as f:
        for i, line in enumerate(f):
            try:
                data = json.loads(line)
                
                # Join sentences for the full article (input)
                # The paper notes these are long, so we truncate to fit context limits
                full_article = " ".join(data['article_text'])
                
                # The abstract is the 'description' (few-shot ground truth)
                # We handle cases where the key might be missing or empty
                abstract_list = data.get('abstract_text', [])
                abstract = " ".join(abstract_list).replace("<S>", "").replace("</S>", "").strip()
                
                # Create the Document object
                doc = Document(
                    page_content=full_article[:6000], # ~1.5k-2k tokens approx
                    metadata={
                        "article_id": data['article_id'],
                        "dcterms:description": abstract, # The target metadata property
                        "source": "pubmed_validation"
                    }
                )
                batch.append(doc)

                # 2. Check if batch is ready for ingestion
                if len(batch) == batch_size:
                    if vector_db is None:
                        # Initialize index with the first batch
                        vector_db = Neo4jVector.from_documents(
                            batch, ollama_embeddings, url=NEO4J_URL, 
                            username=NEO4J_USER, password=NEO4J_PASSWORD,
                            index_name=INDEX_NAME, node_label="PubMedPaper"
                        )
                    else:
                        # Add to existing index
                        vector_db.add_documents(batch)
                    
                    total_count += len(batch)
                    print(f"Successfully ingested {total_count} documents...")
                    batch = [] # Reset the batch

            except Exception as e:
                print(f"Error processing line {i}: {e}")
                continue

        # 3. Ingest any remaining documents in the final partial batch
        if batch:
            if vector_db:
                vector_db.add_documents(batch)
                total_count += len(batch)
                print(f"Ingested final batch. Total: {total_count}")
        
        return vector_db

vector_db = ingest_pubmed_in_batches(FILEPATH, batch_size=100)
`````

## Define a Neo4j simple structure based on DCAT you want to move from a single node to a connected structure.

Relashioships : (:Paper)-[:HAS_CREATOR]->(:Author), (:Paper)-[:HAS_KEYWORD]->(:Keyword).

```cypher
MATCH (p:PubMedPaper {article_id: 'PMC4810892'})
MERGE (a:Author {name: 'Lennart Busch'})
MERGE (k1:Keyword {name: 'VTE'})
MERGE (k2:Keyword {name: 'India'})
MERGE (t:Theme {name: 'Medicine'})
MERGE (p)-[:HAS_CREATOR]->(a)
MERGE (p)-[:HAS_KEYWORD]->(k1)
MERGE (p)-[:HAS_KEYWORD]->(k2)
MERGE (p)-[:HAS_THEME]->(t);
```

Find a paper using Vector Similarity and then instantly "hop" to its neighbors (Authors, Keywords, Themes) using Cypher—all in one single database call.

`````python
from langchain_community.vectorstores import Neo4jVector

# This query is executed for EACH paper found by the vector search
hybrid_retrieval_query = """
// 1. Find neighbors (Authors, Keywords, Themes)
OPTIONAL MATCH (node)-[:HAS_CREATOR]->(a:Author)
OPTIONAL MATCH (node)-[:HAS_KEYWORD]->(k:Keyword)
OPTIONAL MATCH (node)-[:HAS_THEME]->(t:Theme)

// 2. Aggregate findings
WITH node, score, 
     collect(DISTINCT a.name) AS authors, 
     collect(DISTINCT k.name) AS keywords, 
     t.name AS theme

// 3. Format into a single string for the prompt
RETURN 
    "Article Content: " + node.text + 
    "\n[Graph Metadata]" +
    "\n- Creators: " + (CASE WHEN size(authors) > 0 THEN REDUCE(s = "", x IN authors | s + (CASE WHEN s = "" THEN "" ELSE ", " END) + x) ELSE "Data not yet extracted" END) +
    "\n- Keywords: " + (CASE WHEN size(keywords) > 0 THEN REDUCE(s = "", x IN keywords | s + (CASE WHEN s = "" THEN "" ELSE ", " END) + x) ELSE "Data not yet extracted" END) +
    "\n- Theme: " + coalesce(theme, "Uncategorized") AS text, 
    score, 
    node {.*, authors: authors, keywords: keywords, theme: theme} AS metadata
"""
# Initialize the retriever with the custom query
vector_db = Neo4jVector.from_existing_index(
    ollama_embeddings,
    url=NEO4J_URL,
    username=NEO4J_USER,
    password=NEO4J_PASSWORD,
    index_name=INDEX_NAME,
    retrieval_query=hybrid_retrieval_query  # This is where the magic happens
)
`````

`````python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnablePassthrough
from langchain_ollama import ChatOllama
import re
# 1. Define the Graph-Enriched Template
# This matches the DCAT properties (dcterms:description) from the paper
template = """
Extract metadata from the article snippet.
Return ONLY raw JSON. No markdown, no preamble.

STRICT THEME LIST (Choose the best fit):
[Medicine, Biology, Biotechnology, Chemistry, Public Health, Clinical Trials]

JSON Structure:
{{
  "creators": [],
  "keywords": [],
  "theme": ""
}}

Article: {text}
"""

prompt = ChatPromptTemplate.from_template(template)

llm = ChatOllama(model="llama3", temperature=0)


def repair_and_parse_json(raw_output):
    # Use regex to find the first '{' and the last '}'
    match = re.search(r'\{.*\}', raw_output, re.DOTALL)
    if match:
        json_str = match.group(0)
        # Remove any potential comments inside the JSON like (since no authors...)
        json_str = re.sub(r'\(.*?\)', '', json_str)
        return json.loads(json_str)
    raise ValueError("No JSON object found in LLM output")


# 2. Helper function to format the Graph Data into a string for the prompt
def format_graph_docs(docs):
    formatted = []
    for i, doc in enumerate(docs):
        # Using the custom metadata we retrieved in the previous step
        ctx = (
            f"Example {i+1}:\n"
            f"- Title/Content Snippet: {doc.page_content[:200]}...\n"
            f"- Authors: {', '.join(doc.metadata.get('authors', ['Unknown']))}\n"
            f"- Keywords: {', '.join(doc.metadata.get('keywords', ['None']))}\n"
            f"- Existing Description: {doc.metadata.get('dcterms:description', 'No abstract available')}\n"
        )
        formatted.append(ctx)
    return "\n\n".join(formatted)

# 3. Create the Hybrid Chain
# Note: 'vector_db' here is the one initialized with the 'retrieval_query' from the previous step
chain = (
    {
        "examples": vector_db.as_retriever(search_kwargs={"k": 3}) | format_graph_docs,
        "input_text": RunnablePassthrough(),
        "k": lambda x: 3
    }
    | prompt
    | llm # Your Gemini or Ollama instance
    | StrOutputParser()
)

# 4. Execute
new_paper = "Recent studies on the impact of lifestyle on VTE recurrence in Indian populations..."
result = chain.invoke(new_paper)
print(f"Generated Metadata: {result}")
`````

## Extraction and linkage

Since the data is not annotaded, make a script to terates through the ingested PubMed papers, use a LLM to extract DCAT entities, and "wires" them into Neo4j.

`````python
import json
from langchain_community.graphs import Neo4jGraph
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import JsonOutputParser

# 1. Initialize Graph and LLM
# Avoid APOC plugin error message
graph = Neo4jGraph(url=NEO4J_URL, username=NEO4J_USER, password=NEO4J_PASSWORD,refresh_schema=False)

# 2. Define Extraction Chain
extract_prompt = ChatPromptTemplate.from_template("""
Extract metadata from the article snippet.
Return ONLY raw JSON. No markdown code blocks, no preamble, and no conversational text.
If authors are missing, use an empty list [].

Strict JSON structure:
{{
  "creators": ["Name1", "Name2"],
  "keywords": ["Tag1", "Tag2"],
  "theme": "Category"
}}

Article: {text}
""")

# Note: Use your existing llm instance (Gemini/Ollama)
extraction_chain = extract_prompt | llm | StrOutputParser()

def build_graph_from_nodes(limit=50):
    fetch_query = """
        MATCH (p:PubMedPaper) 
        WHERE NOT (p)-[:HAS_KEYWORD]->() 
        RETURN p.article_id AS id, p.text AS text 
        LIMIT $limit
    """
    papers = graph.query(fetch_query, {"limit": limit})

    if not papers:
        print("No papers found that need processing.")
        return

    for paper in papers:
        paper_id = paper['id']
        # Use first 2000 chars to avoid token limits
        content = paper['text'][:2000] if paper['text'] else ""
        
        print(f"Extracting entities for: {paper_id}...")
        
        try:
            # 1. Get the raw string from LLM
            raw_response = extraction_chain.invoke({"text": content})   

            # 2. Repair and parse the JSON
            metadata = repair_and_parse_json(raw_response)         

            # 3. Create the Graph bridges
            link_query = """
            MATCH (p:PubMedPaper {article_id: $id})
            FOREACH (name IN $creators |
                MERGE (a:Author {name: name})
                MERGE (p)-[:HAS_CREATOR]->(a)
            )
            FOREACH (kw IN $keywords |
                MERGE (k:Keyword {name: kw})
                MERGE (p)-[:HAS_KEYWORD]->(k)
            )
            MERGE (t:Theme {name: $theme})
            MERGE (p)-[:HAS_THEME]->(t)
            SET p.status = 'completed'
            """
            
            graph.query(link_query, {
                "id": paper_id,
                "creators": metadata.get('creators', []),
                "keywords": metadata.get('keywords', []),
                "theme": metadata.get('theme', 'General Science')
            })
            
        except Exception as e:
            print(f"Skipping {paper_id} due to error: {e}")

# RUN THIS FIRST to fill your graph
build_graph_from_nodes(limit=10)
`````

## Validate

`````python
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

def calculate_similarity(text1, text2):
    # Get embeddings for both texts
    emb1 = ollama_embeddings.embed_query(text1)
    emb2 = ollama_embeddings.embed_query(text2)
    # Calculate cosine similarity
    return cosine_similarity([emb1], [emb2])[0][0]

def validate_graph_rag(test_limit=5):
    # 1. Fetch papers that are 'completed' (have graph data)
    test_papers = graph.query("""
        MATCH (p:PubMedPaper {status: 'completed'}) 
        RETURN p.article_id AS id, p.text AS original_text 
        LIMIT $limit
    """, {"limit": test_limit})

    results = []
    
    for paper in test_papers:
        print(f"Validating {paper['id']}...")
        
        # 2. Generate new description using your GraphRAG chain
        generated_desc = hybrid_chain.invoke(paper['original_text'][:1000])
        
        # 3. Compare to original
        score = calculate_similarity(paper['original_text'], generated_desc)
        
        results.append({
            "id": paper['id'],
            "score": score,
            "generated": generated_desc[:100] + "..."
        })

    # Calculate average performance
    avg_score = sum(r['score'] for r in results) / len(results)
    print(f"\n--- Validation Complete ---")
    print(f"Average Semantic Match: {avg_score:.2%}")
    return results

# Run validation on 5 nodes
validation_report = validate_graph_rag(5)
`````

### Generate JSON

`````python
viz_query = """
MATCH (p:PubMedPaper)
OPTIONAL MATCH (p)-[r]->(m)
RETURN p.article_id AS source_id, 
       p.status AS status,
       labels(m)[0] AS target_type, 
       m.name AS target_name,
       type(r) AS rel_type
LIMIT 500
"""

results = graph.query(viz_query)

nodes = {}
links = []

for row in results:
    # 1. Add/Update Paper Node
    # Group 1 = Processed (Green), Group 0 = Pending (Red)
    status_group = 1 if row['status'] == 'completed' else 0
    nodes[row['source_id']] = {
        "id": row['source_id'], 
        "group": status_group, 
        "label": f"Paper ({row['status']})"
    }
    
    # 2. Add Target Node (if it exists)
    if row['target_name']:
        target_id = f"{row['target_type']}_{row['target_name']}"
        if target_id not in nodes:
            nodes[target_id] = {
                "id": target_id, 
                "name": row['target_name'], 
                "group": 2, # Entities (Blue/Yellow)
                "label": row['target_type']
            }
        
        # 3. Add Link
        # Add Link with the Relationship Type
        links.append({
            "source": row['source_id'], 
            "target": target_id,
            "rel_type": row['rel_type'] # Added this!
        })

with open('graph_data.json', 'w') as f:
    json.dump({"nodes": list(nodes.values()), "links": links}, f)
`````

## Build graph from nodes

`````python
from tqdm import tqdm

from tqdm import tqdm

def full_background_processor(total_to_process=1000, batch_size=20):
    # Setup progress bar
    pbar = tqdm(total=total_to_process, desc="GraphRAG Construction", unit="paper")
    
    for _ in range(0, total_to_process, batch_size):
        try:
            # We target 'pending' nodes specifically now
            build_graph_from_nodes(limit=batch_size) 
            pbar.update(batch_size)
        except Exception as e:
            print(f"Batch failed: {e}")
            
    pbar.close()
    print("Batch processing complete. Check your 3D Viz!")

full_background_processor(5)
`````
